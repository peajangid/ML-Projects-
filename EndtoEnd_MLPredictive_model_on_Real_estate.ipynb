**Using log transformation for non normal columns**

```python
for feature in continuous_feature:
    df_copy = df.copy()

    # Check for zero or negative values
    if (df_copy[feature] <= 0).any():
        # Replace zero or negative values with NaN or a small positive value
        df_copy[feature] = np.where(df_copy[feature] > 0, np.log(df_copy[feature]), np.nan)
    else:
        # Apply log transformation
        df_copy[feature] = np.log(df_copy[feature])

    # Ensure 'SalePrice' is transformed similarly
    df_copy['SalePrice'] = np.log(df_copy['SalePrice'])

    # Plot the relationship
    plt.scatter(df_copy[feature], df_copy['SalePrice'])
    plt.title(f'{feature} vs SalePrice')
    plt.xlabel(feature)
    plt.ylabel('Sale Price')
    plt.show()







```

### Categorical variables
removing the rare categorical features

```python
cat_features = [feature for feature in df.columns if df[feature].dtypes == 'O']
cat_features

for feature in cat_features:
    temp=df.groupby(feature)['SalePrice'].count()/len(df)
    temp_df=temp[temp>0.01].index
    df[feature]=np.where(df[feature].isin(temp_df),df[feature],'Rare_var')


```

```python
mappings = {}
for feature in cat_features:
  labels_ordered = df.groupby([feature])['SalePrice'].mean().sort_values().index
  labels_ordered = {k:i for i,k in enumerate(labels_ordered,0)}
  mappings[feature] = labels_ordered
  df[feature] = df[feature].map(labels_ordered)

df
```

### Test data feature engineering
1. Missing value imputation
2. Handling temporal variables
3. scaling the numerical features

```python
# Imputing missing values for categorical columns
cat_features = [feature for feature in test_df.columns if test_df[feature].dtypes == 'O' and test_df[feature].isnull().sum()>0]
cat_features

for feature in cat_features:
    test_df[feature] = test_df[feature].fillna('Missing')
```

```python
cat_features = [feature for feature in test_df.columns if test_df[feature].dtypes == 'O']
# For encoding the same features in the test data
for feature in cat_features:
    if feature in mappings:
        # Map test data categories using training mappings
        test_df[feature] = test_df[feature].map(mappings[feature])

        # Identify any new categories in the test data not seen in the training data
        unseen_categories = test_df[feature].isin(mappings[feature].keys()) == False
        new_categories = test_df[feature][unseen_categories].unique()

        # Add the new categories to the mappings with a new encoding
        for new_category in new_categories:
            # Add new category to the mapping with a unique integer
            mappings[feature][new_category] = len(mappings[feature])

        # Re-map the test data with updated mappings
        test_df[feature] = test_df[feature].map(mappings[feature])

        # Handle any remaining NaN values (categories still not found)
        test_df[feature] = test_df[feature].fillna(-1).astype(int)
```

### **Feature Selection**

```python
from sklearn.linear_model import Lasso
from sklearn.feature_selection import SelectFromModel
```

```python
data = pd.read_csv('X_train.csv')

y_train = data['SalePrice']
X_train = data.drop(['SalePrice'],axis = 1)
```

```python
# Using lasso regressionmodel to select featues
# The higher the aplha(Penealyt term) the less features will be selected
feature_selection_model = SelectFromModel(Lasso(alpha=0.005,random_state=0))
feature_selection_model.fit(X_train,y_train)

feature_selection_model.get_support()
```

```python
# Selecting the features
selected_features = X_train.columns[(feature_selection_model.get_support())]

# Printing stats
print('Total features:',X_train.shape[1])
print('Total Selected features:', len(selected_features))
print('Total Features whose coeffiecient shrak to 0:', np.sum(feature_selection_model.estimator_.coef_==0))
```
